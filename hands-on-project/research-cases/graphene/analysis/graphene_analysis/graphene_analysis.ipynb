# Density of States (DOS) Visualization


The Density of States (DOS) represents the number of
available electronic states at different energy levels.


The DOS calculation is obtained from Quantum ESPRESSO
using the `dos.x` post-processing tool.


````python
dos_file = "../output/graphene_dos.dat"


if os.path.exists(dos_file):

    dos = np.loadtxt(dos_file)

    energy = dos[:,0]
    density = dos[:,1]


    plt.figure(figsize=(7,4))


    plt.plot(
        energy,
        density
    )


    plt.xlabel("Energy (eV)")
    plt.ylabel("DOS")


    plt.title(
        "Graphene Density of States"
    )


    plt.grid()

    plt.show()


else:

    print("DOS data is not available yet")
````

# Band Structure Visualization


The band structure describes the relationship between
electron energy and crystal momentum.


For graphene, the calculation is performed along the
high-symmetry path: Γ → K → M → Γ



````python
band_file = "../output/graphene_bands.dat.gnu"


if os.path.exists(band_file):

    data = np.loadtxt(band_file)


    k = np.unique(data[:,0])


    bands = np.reshape(
        data[:,1],
        (-1,len(k))
    )


    plt.figure(figsize=(7,5))


    # Plot all electronic bands
    for band in range(len(bands)):

        plt.plot(
            k,
            bands[band,:],
            linewidth=1
        )


    # High symmetry points
    plt.xticks(
        ticks=[
            0,
            0.6667,
            1,
            1.5774
        ],
        labels=[
            "Γ",
            "K",
            "M",
            "Γ"
        ]
    )


    # Vertical lines at symmetry points
    plt.axvline(
        0.6667,
        linestyle="--",
        linewidth=0.8
    )

    plt.axvline(
        1,
        linestyle="--",
        linewidth=0.8
    )


    # Fermi level reference
    plt.axhline(
        0,
        linestyle="--",
        linewidth=0.8
    )


    plt.xlabel(
        "High Symmetry Path"
    )


    plt.ylabel(
        "Energy (eV)"
    )


    plt.title(
        "Graphene Band Structure"
    )


    plt.grid()

    plt.show()


else:

    print("Band structure data is not available yet")
````

# Electronic Structure Interpretation


The electronic properties of graphene are analyzed by
combining DOS and band structure results.


## DOS Analysis

DOS provides information about:

- Distribution of electronic states
- Available states near the Fermi level
- Electronic contribution from energy regions


## Band Structure Analysis

Band structure provides information about:

- Energy dispersion
- Valence and conduction bands
- High symmetry points in reciprocal space


For graphene, the band structure is expected to show
the characteristic Dirac point near the K point.

# 1. Density of States (DOS) Analysis

Density of States describes the distribution of electronic states as a function of energy.

````python
dos_file = '../output/graphene_dos.dat'

if os.path.exists(dos_file):
    dos = np.loadtxt(dos_file)
    energy = dos[:,0]
    density = dos[:,1]
    print('DOS data loaded')

else:
    print('DOS file not found:')
    print(dos_file)
````

````python
if os.path.exists(dos_file):
    plt.figure(figsize=(6,4))

    plt.plot(
        energy,
        density
    )

    plt.xlabel('Energy (eV)')
    plt.ylabel('DOS')
    plt.title('Graphene Density of States')
    plt.grid()
    plt.show()
````

# 2. Band Structure Analysis

Band structure describes the relationship between electron energy and crystal momentum.

````python
band_file = '../output/graphene_bands.dat.gnu'

if os.path.exists(band_file):
    data = np.loadtxt(band_file)

    k = np.unique(data[:,0])

    bands = np.reshape(
        data[:,1],
        (-1,len(k))
    )

    print('Band data loaded')

else:
    print('Band file not found:')
    print(band_file)
````

````python
if os.path.exists(band_file):
    plt.figure(figsize=(7,5))

    for band in range(len(bands)):
        plt.plot(
            k,
            bands[band,:]
        )

    plt.xlabel('K-point')
    plt.ylabel('Energy (eV)')
    plt.title('Graphene Band Structure')
    plt.grid()
    plt.show()
````
